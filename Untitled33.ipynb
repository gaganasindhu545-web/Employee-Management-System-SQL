```python
# list is used to store the data, it is accessed by using index starts from zero, it is mutable which means it can change,it is used when data has to changes
# In list we can use different datatypes to store the data
# creating a lists
list=["sindhu","hari","deepthi","Nikhila","janu"]
print(list)
list1=[34,56,76,24,36]
print(list1)
list2=["sindhu",34,"hari",56]
print(list2)
list.append('kalyani')
list.remove('janu')
print(list)
print(len(list))
# loops in list
for name in list:
  print(name,end=" ")
# using enumerate it is used when we want indexes+value
print()
for i,val in enumerate(list):
  print(i,val)
```

```python
# tuple it is also used to store the data as like list stores but it is immutable which means in this data is fixed we can't change the data
# tuple also allows different datatypes to store in it.
# creating tuple
tuple=('sindhu','hari','deepthi','kalyani','nikhila')
print(tuple)
# tuple.append('akhila') # this will give error because we cannot change or update the data stored in tuple
# tuple.remove('sindhu') # this will give error because we cannot remove the data stored in tuple
tuple1=('sindhu',34,'hari',56)
print(tuple1)
# creating loops
for i in tuple:
    print(i,end=" ")
print()
# using enumeration
for index,name in enumerate(tuple):
  print(index,name)
```

```python
# dictionary
# dictionary is used to store data in key-value pairs, it allows to access data or values by using keys
# key=word,meaning=value.
# creating simple dictionary
student={
    'name':'sindhu',
    'age':19,
    'college':'Aditya',
    'marks':89
    }
print(student['name'])
# creating lists in dictionary
student1={
    'name':['sindhu','hari','deepthi'],
    'age':[19,15,19],
    'marks':[89,80,86]
}
print(student1['name'])
print(student1['age'])
print(student1['marks'])
# accessing particular student data
print(student1['name'][0],student1['age'][0],student1['marks'][0])
#inserting new data
student['branch']='DS'
print(student)
# deleting data
del student['college']
print(student)
# updating
student['age']=20
print(student)
# get keys
print(student1.keys())
# get values
print(student1.values())
# get key and value
print(student1.items())
# loops
for key,value in student1.items():
  print(key,value)
```

```python
#Set
# set is a unordered collection it is used to store unique values
# In set we can add,remove the data,useful operations are intersection,union,difference
# creating a set
cities={'hyderabad','delhi','mumbai','hyderabad'}
print(cities)
list=['sindhu','hari','sindhu']
updated=set(list)
print(updated)
cities.add('chennai')
cities.remove('delhi')
print(cities)
a={10,20,30,40}
b={30,40,50,60}
print(a|b)  # combine unique elements
print(a&b)  # common elements
print(a-b)  # present in a but not in b
```
